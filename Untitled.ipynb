```python
import pandas as pd

# URLs for each year's rental bond lodgement Excel file
rental_data_urls = {
    2021: "https://www.nsw.gov.au/sites/default/files/noindex/2023-11/Rental-bond-lodgements-year-2021.xlsx",
    2022: "https://www.nsw.gov.au/sites/default/files/noindex/2023-11/RentalBond_Lodgements_Year_2022.xlsx",
    2023: "https://www.nsw.gov.au/sites/default/files/noindex/2024-05/RentalBond_Lodgements_Year_2023.xlsx",
    2024: "https://www.nsw.gov.au/sites/default/files/noindex/2025-01/rental-bond-lodgements-year-2024_1.xlsx"
}

dfs = []

for year, url in rental_data_urls.items():
    print(f"Loading data for {year}...")
    # Load first sheet without specifying sheet name
    df = pd.read_excel(url, sheet_name=0, skiprows=2)  # skiprows=2 to remove header rows if needed
    
    # Clean up empty rows and reset index
    df.dropna(how='all', inplace=True)
    df.reset_index(drop=True, inplace=True)
    dfs.append(df)

# Combine all dataframes into one
df_all_years = pd.concat(dfs, ignore_index=True)

print(df_all_years.head())
```

```python
def load_2025_monthly_data(url, month_name):
    df = pd.read_excel(url, sheet_name=0, skiprows=2)
    df.dropna(how='all', inplace=True)
    df.reset_index(drop=True, inplace=True)
    return df

# URLs for Jan–Apr 2025
monthly_urls = {
    "January": "https://www.nsw.gov.au/sites/default/files/noindex/2025-02/rental-bond-lodgements-data-january-2025.xlsx",
    "February": "https://www.nsw.gov.au/sites/default/files/noindex/2025-03/rental-bond-lodgement-data-february-2025.xlsx",
    "March": "https://www.nsw.gov.au/sites/default/files/noindex/2025-04/rental-bond-lodgement-data-march-2025.xlsx",
    "April": "https://www.nsw.gov.au/sites/default/files/noindex/2025-05/rental-bond-lodgements-april-2025.xlsx"
}

# Load and combine all 2025 monthly data
df_2025_monthly = pd.concat(
    [load_2025_monthly_data(url, month) for month, url in monthly_urls.items()],
    ignore_index=True
)

# Append to existing df_all_years (assumes it's already loaded)
df_all_years = pd.concat([df_all_years, df_2025_monthly], ignore_index=True)
```

```python
df_all_years['Weekly Rent'] = pd.to_numeric(df_all_years['Weekly Rent'], errors='coerce')
median_rent = df_all_years['Weekly Rent'].median()
df_all_years['Weekly Rent'] = df_all_years['Weekly Rent'].fillna(median_rent)
```

```python
df_all_years['Bedrooms'] = df_all_years['Bedrooms'].replace('U', pd.NA)
df_all_years['Bedrooms'] = pd.to_numeric(df_all_years['Bedrooms'], errors='coerce')
```

```python
mode_bedrooms = df_all_years['Bedrooms'].mode()[0]
df_all_years['Bedrooms'] = df_all_years['Bedrooms'].fillna(mode_bedrooms)
```

```python
df_all_years.loc[:,'Pricing_segment'] = df_all_years.loc[:,'Weekly Rent'].apply(
    lambda x: 'expensive' if x > (median_rent + 200) else ('affordable' if x < (median_rent + 200) and x > median_rent else 'Inexpensive')
)
```

```python
def segment_postcode(postcode):
    if 2000 <= postcode <= 2234:
        return 'Sydney Metro'
    elif 2250 <= postcode <= 2490:
        return 'Regional NSW'
    elif 2619 <= postcode <= 2898 or 2921 <= postcode <= 2999:
        return 'Remote NSW'
    else:
        return 'Other NSW'
    
```

```python
df_all_years['Postcode_segment'] = df_all_years.loc[:,'Postcode'].apply(segment_postcode)
```
